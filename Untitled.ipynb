```python
import icsd
import numpy as np
import spike_tools as st
import nn_tool as nt
import parameters as pm
import pickle
import seaborn as sns
import quantities as pq
import neo.core

# import pandas as pd
# used icsd.py to get raw csd. If you use estimated_csd or elephant,
# which gives you the output of estimated_csd, you will have filtered csd
# instead of raw one, which is different from what your supervisor asked you
# to calculate. But maybe it's not bad to study how to get estimated_csd.
# for this one, I used traditional CSD filtered with hamming three point
# method.

'''
This is for a single animal data
'''
file_path = '/home/bhc/OneDrive/Work/PhD/projects/zwicker_tone_2018_2021' \
            '/data/dataset/'

animal = 'cr50_190724'

d_lfp = file_path + 'lfp_fil/' + animal + '_lfp_fil.hdf5'
d_spk = file_path + 'tsp/' + animal + '_tsp.hdf5'
sr_lfp = 1061.5 * pq.Hz

with open('/home/bhc/OneDrive/Work/PhD/projects/zwicker_tone_2018_2021/data'
          '/dataset/nn_lfp_avg/cr50_190724_nn_3_1_lfp_a.pk', 'rb') as dat:
    datas = pickle.load(dat)

la = np.array(datas['la']['nn_3_1']) * pq.mV
lfp_data = neo.core.AnalogSignal(la.T, sampling_rate=sr_lfp)
coord = np.arange(0, 1501, 100) * pq.micrometer
sigma = 1 / (100 * pq.micrometer)
'''
I don't know how this works, but the  approximation of raw csd is (V[z-h] 
-2 * V[z]**2 + V[z+h])/h ** 2, which removes the conductance out of the 
formula. So instead of using the default sigma, I changed the value to 
1/h, because in icsd.py, h is divided only one time.
'''
# input args for standard CSD function in iCSD.py
std_input = {
    'lfp': lfp_data,
    'coord_electrode': coord,
    'sigma': sigma,
    'f_type': 'hamming',
    'f_order': 3,
    }

std1 = icsd.estimate_csd(**std_input)
r_csd = np.transpose(std1[0].magnitude) * std1[0].units
f_csd = np.transpose(std1[1].magnitude) * std1[0].units
r_csd.units = pq.mV/(pq.mm**2)
f_csd.units = pq.mV/(pq.mm**2)
```

```python
import matplotlib.pyplot as plt
from matplotlib import colors
from matplotlib import ticker
import matplotlib.cm as cm

def make_plot(xx, yy, zz, title='True CSD', cmap=cm.bwr):
    fig = plt.figure(figsize=(7, 7))
    ax = plt.subplot(111)
    ax.set_aspect('equal')
#     t_max = np.max(np.abs(zz))
#     levels = np.linspace(-1 * t_max, t_max, 32)
    im = ax.contourf(xx, yy, zz, cmap=cmap)
    ax.set_xlabel('X (mm)')
    ax.set_ylabel('Y (mm)')
    ax.set_title(title)
#     ticks = np.linspace(-1 * t_max, t_max, 3, endpoint=True)
    plt.colorbar(im, orientation='horizontal', format='%.2f', ticks=ticks)
    return ax
make_plot(t_c, coord, r_csd, title='True CSD', cmap=cm.bwr)
```
